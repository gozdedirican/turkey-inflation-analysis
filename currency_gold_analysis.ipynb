```
import pandas as pd
import matplotlib.pyplot as plt
from prophet import Prophet
# CSV dosyasını oku
df = pd.read_csv('currency_gold.csv')
df['Date'] = pd.to_datetime(df['Date'])
# Dolar grafiği
plt.figure(figsize=(10,5))
plt.plot(df['Date'], df['USD_TRY'], marker='o', label='USD/TRY')
plt.plot(df['Date'], df['Gold'], marker='x', label='Gold')
plt.title('USD/TRY and Gold Prices')
plt.xlabel('Date')
plt.ylabel('Price')
plt.legend()
plt.grid(True)
plt.show()
# Basit tahmin için Prophet kullanabiliriz
df_prophet = df[['Date','USD_TRY']].rename(columns={'Date':'ds','USD_TRY':'y'})
model = Prophet()
model.fit(df_prophet)
future = model.make_future_dataframe(periods=5)
forecast = model.predict(future)
print(forecast[['ds','yhat','yhat_lower','yhat_upper']].tail())
```
