```
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('turkey_cpi.csv')
df['Date'] = pd.to_datetime(df['Date'])
df.set_index('Date', inplace=True)
plt.figure(figsize=(10,5))
plt.plot(df.index, df['CPI'], marker='o')
plt.title('Turkey CPI Over Time')
plt.ylabel('CPI')
plt.grid(True)
plt.show()
df['YoY_pct_change'] = df['CPI'].pct_change(periods=12) * 100
print(df)
```
