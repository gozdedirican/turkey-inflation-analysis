```
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
# CSV dosyasını oku
df = pd.read_csv('sales_data.csv')
# SQLite veritabanı oluştur
conn = sqlite3.connect('sales_data.db')
df.to_sql('Sales', conn, if_exists='replace', index=False)
# SQL sorguları
query_total_sales = 'SELECT Product, SUM(Quantity*Price) as TotalSales FROM Sales GROUP BY Product'
total_sales = pd.read_sql_query(query_total_sales, conn)
print(total_sales)
# Grafik
plt.figure(figsize=(8,5))
plt.bar(total_sales['Product'], total_sales['TotalSales'], color=['blue','green','orange'])
plt.title('Total Sales by Product')
plt.ylabel('Total Sales')
plt.show()
```
